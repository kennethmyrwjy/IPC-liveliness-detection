```python
import cv2
import matplotlib.pyplot as plt

def check_if_real(image_path):
    """
    Simulates a liveness check.
    """
    print(f"Analyzing '{image_path}' for liveness...")
    is_spoof = "fake" in image_path.lower()
    return "SPOOF" if is_spoof else "REAL"

def load_and_prepare_image(image_path):
    """
    Loads, rotates, resizes, and converts the image to RGB.
    """
    image = cv2.imread(image_path)
    if image is None:
        raise FileNotFoundError(f"Image not found: {image_path}")

    # Rotate 90 degrees clockwise
    image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

    # Resize and convert to RGB for matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image_rgb = cv2.resize(image_rgb, (500, 800))
    return image_rgb

# --- Test Cases ---
spoof_image_path = "dataset/train/fake/IMG_1002.JPG"
real_image_path = "dataset/train/real/kenasli_real.jpg"

# Analyze both images
result_spoof = check_if_real(spoof_image_path)
result_real = check_if_real(real_image_path)

# Load and prepare both images
image_spoof = load_and_prepare_image(spoof_image_path)
image_real = load_and_prepare_image(real_image_path)

# --- Display Side by Side ---
plt.figure(figsize=(12, 9))

# Spoof Image
plt.subplot(1, 2, 1)
plt.imshow(image_spoof)
plt.title(f"Liveness: {result_spoof}", fontsize=14, color='red')
plt.axis('off')

# Real Image
plt.subplot(1, 2, 2)
plt.imshow(image_real)
plt.title(f"Liveness: {result_real}", fontsize=14, color='green')
plt.axis('off')

plt.tight_layout()
plt.show()
```
